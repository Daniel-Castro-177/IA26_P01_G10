### Contexto

A procura de comida local tem vindo a aumentar na UE, mas os pequenos produtores têm custos de logística elevados. Uma plataforma que ligue produtores, clientes e centros de recolha pode tornar a distribuição local mais eficiente, sendo o principal desafio o planeamento das rotas de recolha e entrega.

### Objetivo

Desenvolver um agente que escolha rotas de recolha e entrega para os veículos dos centros de recolha, respeitando as janelas temporais (disponibilidade no produtor e entrega ao cliente) e a capacidade dos veículos. Os dados são sintéticos (CSV) e cobrem o distrito de Braga e a Póvoa de Varzim.
